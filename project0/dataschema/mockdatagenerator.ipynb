```python
# goal is to generate each excel file with min 8-24 records number of records refers to number of cars washed in a day
# each excel file name is in format washdata_ddMMyyyy.xlsx where ddMMyyyy is the date of the records in the file and new day follows same cycle 
# jan 1 2023 to dec 31 2023 so total 365 files to be generated
# each file will have records from user_id 1 to n (n = number of records in that day) next day will continue from n+1 to m (m = number of records in next day)

# doc1 user_id end will continue to doc2 
# Each doc ( represents one day with date in format dd/MM/yyyy) record contains the following fields:
# user_id : incremental from 1 to as many records as we want
# phone : a phone number Australian format 0XXX XXX XXX
# wash_service : randomly choose from ['Deluxe : 70 ', 'Xtreme : 100', 'Hand Polish : 120', 'Super Wash : 200', 'Detail : 300'] the count of service is in order of price low to high 
# entry_time : 8:00 AM to 6:00 PM randomly generated datetime today incremental with user_id 
# out_time : entry_time + random duration between 45 minutes to 2 hours
# payment_method : randomly choose from ['Credit Card', 'Cash']
# amount_paid : based on wash_service chosen amount paid is the price of the service refer above
# comments : optional field if its vourcher used or staff free wash {voucher : 'VOUCHER CODE XYZ', staff_free : 'Staff Free Wash'} else "regular wash"

# Example record:
# user_id : 1
# phone : 0412 345 678
# wash_service : Deluxe
# entry_time : 01/01/2023 08:15 AM
# out_time : 01/01/2023 09:00 AM
# payment_method : Credit Card
# amount_paid : 70
# comments : regular wash
# num_records = 15  # number of records for the day



```

```python
# lets begin 
faker = Faker('en_AU')  # Australian locale

start_date = datetime(2023, 1, 1)
end_date = datetime(2023, 12, 31)
current_user_id = 1
date_generated = [start_date + timedelta(days=x) for x in range(0, (end_date-start_date).days + 1)]
for single_date in date_generated:
    num_records = random.randint(8, 24)
    daily_records = []
    entry_time_base = single_date.replace(hour=8, minute=0, second=0)
    for i in range(num_records):
        user_id = current_user_id
        faker = Faker('en_AU')

        phone = "04" + "".join(str(random.randint(0, 9)) for _ in range(8))
        wash_service, amount_paid = random.choice([
            ('Deluxe', 70), 
            ('Xtreme', 100), 
            ('Hand Polish', 120), 
            ('Super Wash', 200), 
            ('Detail', 300)
        ])
        entry_time = entry_time_base + timedelta(minutes=random.randint(0, 600))  # between 8:00 AM to 6:00 PM
        out_time = entry_time + timedelta(minutes=random.randint(45, 120))  # wash duration between 45 mins to 2 hours
        payment_method = random.choice(['Credit Card', 'Cash'])
        comments = "regular wash"
        if random.random() < 0.2:  # 20% chance of voucher or staff free wash
            if random.random() < 0.5:
                comments = f"voucher : 'VOUCHER CODE {faker.bothify(text='???-#####')}'"
            else:
                comments = "staff_free : 'Staff Free Wash'"
        
        daily_records.append({
            "user_id": user_id,
            "phone": phone,
            "wash_service": wash_service,
            "entry_time": entry_time.strftime("%d/%m/%Y %I:%M %p"),
            "out_time": out_time.strftime("%d/%m/%Y %I:%M %p"),
            "payment_method": payment_method,
            "amount_paid": amount_paid,
            "comments": comments
        })
        current_user_id += 1
    
    daily_df = pd.DataFrame(daily_records)
    file_name = f"washdata_{single_date.strftime('%d%m%Y')}.xlsx"
    daily_df.to_excel(file_name, index=False)
    print(f"Generated file: {file_name} with {num_records} records.")
```
