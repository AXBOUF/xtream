```python
import os
import glob

MONTHS = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']

def get_file_paths():
    files = []
    for month in MONTHS:
        pattern = os.path.join(os.getcwd(), '2022', month, '*.xlsx')
        files.extend(glob.glob(pattern))
    return files
```

```python
for path in get_file_paths():
    print(path)  # or pass to another function
```
